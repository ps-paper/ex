```python
import csv
fails = []
with open("..\\data\\08_press.csv") as f:
    reader = csv.reader(f)
    header = next(reader)
    for row in reader:
        if row[5] == "1":
            fails.append(row)
with open("..\\data\\08_fail.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    for row in fails:
        writer.writerow(row)
with open("..\\data\\08_fail.csv", "r", newline="") as f:
    print (f.read())
    f.close()
```
