```python
def min_max(values):
    return min(values), max(values)
result = min_max([75.3, 88.0, 49.1])
print(result)
```

```python
def sensor(value):
    return min(value), max(value), sum(value)/len(value)
low, high, avg = sensor([70,90,80])
print(low, high, avg)
```

```python
def stat(status):
    return max(status), min(status), sum(status)/len(status)
lo, hi, mi = stat([10,50,80])
print(lo, hi, mi)
```

```python
def stats(value):
    return max(value), min(value), sum(value)/len(value)
lo, hi, total = stats([10,20,30])
print(lo, hi, total)
```

```python
def status(temp, limit=90):
    if temp > limit:
        print("경고")
    else:
        print("정상")
status(78)
status(95)
status(50, limit=40)
```

```python
def sensor(li):
    return sum(li) / len(li)
def status(avg, limit=90):
    if avg > limit:
        return "경고"
    else:
        return "정상"
avg = sensor([78, 92, 85])
print(avg,status(avg))
```
