```python
score = int(input())

while score != nt:
    nt = int(input())
    if score > nt:
        print("더 높습니다 다시 입력하세요")
    elif score < nt:
        print("더 낮습니다 다시 입력하세요")
    elif score == nt:
        break
print("정답입니다!!")    
```
