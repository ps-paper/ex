# Chapter 6. 문자열


```python
# 1. 답(5)
s = "hello"
t = "my python"
print(s,"!",t[2:])
# 'my '는 3개의 스택을 차지하고 있기에 t[2:]를 해야 my 삭제 후 python 잇기 가능
```

```python
# 2. 답(3)
# data.islower()는 문자열을 소문자로 변환해 주는 함수가 아니라, 
# 문자열 내의 알파벳이 모두 소문자로만 이루어져 있는지 판별하여 True 또는 False 
# 값을 반환하는 함수이다.
```

```python
# 3. 답(3)
print("abc\bd")
# \b는 백스페이스 키이다.
```

```python
# 4. 답(5)
value_1 = '5'
value_2 = '5-2-10-10'.split('-')[-1]

print(int(value_1)*3+float(value_2))
# value_2에서 '-'를 기준으로 나누니 5,2,10,10의 리스트가 만들어 지고, 
# -1번째니 문자열10이 된다. 따라서 단순계산하면 25.0이 최종 결과값이다.
```

```python
# 5. 답(1)
"H-e-l-l-o-P-y-t-h-o-n"[::2]
# [::2]는 끝까지 두칸 씩 건너뛰라는 의미이며, 0,2,4,6 --- 의 값만 가져온다.
```

```python
# 6. 답(?)
course_name = 'This is New AI World'

for i in course_name:
    if i == 'world':
        i = course_name.lower()

print(course_name)

```

```python
# 8. 답(4)
sentence = "hello, my name is python?!"
print(sentence[0]+sentence[5]+sentence[8:11])
# 0에서 'h', 5에서',' , 8:11에서 'y n'을 얻었다.
```
