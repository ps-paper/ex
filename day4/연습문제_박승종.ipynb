```python
# 1 (답: 3) zip과 내적 연산을 따라가면 아래와 같은 답이 나온다. 행렬 문제인 거 같다.
matrix_a = [[1,2,3,5], [1,2,3,4]]
matrix_b = [[40,40], [30,30], [20,20], [10,10]]
print([[sum([t[0]*t[1] for t in zip(i,j)]) for j in zip(*matrix_b)] for i in matrix_a])
```
