```python
names = ["../data/09_ict_inspection.csv",
         "../data/09_ict_inspection_dirty.csv",
         "../data/09_없는파일.csv"]
done = 0
for name in names:
    try:
        with open(name, encoding="utf-8") as f:
            f.readlines()
        done += 1
    except FileNotFoundError:
        continue   # 없는 파일은 건너뜀
print("처리한 파일:", done)  
```
